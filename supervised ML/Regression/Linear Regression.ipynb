```python
x = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

we spell the data in x and y

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=1)
```

also div into training and testing datan

```python
def Entermodel(model):
  model.fit(x_train,y_train)
  y_pred = model.predict(x_test)
  from sklearn.metrics import r2_score
  r2= r2_score(y_test,y_pred)
  train= model.score(x_train,y_train)
  test= model.score(x_test,y_test)
  print(f"R2 score is {r2}")
  print()
  print(f"train score is {train}")
  print(f"test score is {test}")
  return r2
```

# Redefining the function without accuracy_score (which is only for classification)

```python
from sklearn.linear_model import LinearRegression
li = Entermodel(LinearRegression())
```

now the linear regression given me the traing acc 96% with testing acc 76%.


```python
from sklearn.linear_model import Lasso
L1 = Lasso(alpha=10)
L1.fit(x_train,y_train)
y_pred = L1.predict(x_test)
from sklearn.metrics import r2_score
r2= r2_score(y_test,y_pred)
train= L1.score(x_train,y_train)
test= L1.score(x_test,y_test)
print(f"R2 score is {r2}")
print()
print(f"train score is {train}")
print(f"test score is {test}")
```

```python
from sklearn.linear_model import Ridge
L2 = Ridge(alpha=10)
L2.fit(x_train,y_train)
y_pred = L2.predict(x_test)
from sklearn.metrics import r2_score
r2= r2_score(y_test,y_pred)
train= L2.score(x_train,y_train)
test= L2.score(x_test,y_test)
print(f"R2 score is {r2}")
print()
print(f"train score is {train}")
print(f"test score is {test}")
```
